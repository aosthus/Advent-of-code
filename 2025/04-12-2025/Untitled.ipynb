```python
raw_data = open("./data.txt","r") 
splitted_text = raw_data.read().split('\n')
raw_data.close()
splitted_text.insert(0, '.'*len(splitted_text[0]))
splitted_text.append('.'*len(splitted_text[-1]))
```

```python
num_accessibble_rolls = 0

for i, meter in enumerate(splitted_text[1:-1]):
    meter = '.'+meter+'.'
    print(i)
    for n, suspect in enumerate(meter[1:-1]):
        num_adj_rolls = -1
        if suspect == '@':
            for j in splitted_text[i-1:i+2]:
                print("line to compare with:", j)
                for k in meter[n-1:n+2]:
                    if splitted_text[j][k]=='@':
                        num_adj_rolls += 1
        if num_adj_rolls < 4 and num_adj_rolls != -1:
            num_accessibble_rolls += 1
```
