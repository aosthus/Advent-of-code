```python
raw_data = open("./data.txt","r") 
splitted_text = raw_data.read().split('\n')
raw_data.close()
```

```python
sum_of_jolts = 0

for bank in splitted_text:
    first_digit = max(bank)
    first_slice, second_slice = bank.split(first_digit, 1)[0], bank.split(first_digit, 1)[1]
    if second_slice == '':
        second_digit = max(first_slice)
        first_digit, second_digit = second_digit, first_digit
    else:
        second_digit = max(second_slice)
    sum_of_jolts += int(first_digit+second_digit)
    
sum_of_jolts
```
