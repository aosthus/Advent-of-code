```python
raw_data = open("./data.txt","r") 
splitted_text = raw_data.read().split('\n')
raw_data.close()
```

```python
start_pos = 50
target = 0
num_of_hits = 0

for line in splitted_text:
    direction = line[0]
    rot_len = int(line[1:])
    if line[0] == "L":
        start_pos += rot_len
    elif line[0] == "R":
        start_pos += -rot_len

    if start_pos % 100 == target:
        num_of_hits += 1
```
