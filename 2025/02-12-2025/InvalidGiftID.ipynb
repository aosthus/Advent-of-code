```python
raw_data = open("./data.txt","r") 
splitted_text = raw_data.read().split(',')
raw_data.close()
```

```python
sum_invalid_ids = 0

for item_range in splitted_text:

    start_id = int(item_range.split("-")[0])
    end_id = int(item_range.split("-")[1])

    for id in range(start_id, end_id+1):

        if len(str(id)) % 2 == 0:
            non_repeat_strings = 0

            for char in range((len(str(id)) // 2 )):
                if str(id)[char] != str(id)[char + (len(str(id)) // 2 )]:
                    non_repeat_strings += 1
            
            if non_repeat_strings == 0:
                sum_invalid_ids += id
```

```python
print("Sum of invalid ids from task 1:", sum_invalid_ids)
```
